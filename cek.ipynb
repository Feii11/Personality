```python

# Baca file Excel
df = pd.read_csv('Data.csv', dtype={'SIC': str})  # Ganti dengan nama file kamu

# Ambil 2 digit pertama dari kolom SIC (pastikan kolom dalam bentuk string)
df['SIC'] = df['SIC'].astype(str).str.zfill(4)
df['SIC_2digit'] = df['SIC'].astype(str).str[:2]

# Buat set dari 2 digit yang ada di data
data_sic_2digit = set(df['SIC_2digit'].unique())

# Buat set semua kombinasi SIC 2 digit yang valid (dari 00 sampai 99)
semua_sic_2digit = set(f"{i:02}" for i in range(100))

# Cek mana yang belum ada
tidak_ada = sorted(semua_sic_2digit - data_sic_2digit)

# Tampilkan hasil
print("Jumlah total kode 2 digit yang ditemukan:", len(data_sic_2digit))
print("Kode 2 digit yang belum ada di data:")
print(tidak_ada)
```

```python
# Baca data utama (yang berisi kode SIC)
df = pd.read_csv('Data.csv')  # Ganti dengan nama file kamu

# Pastikan SIC adalah string 4 digit
df['SIC'] = df['SIC'].astype(str).str.zfill(4)

# Ambil 2 digit pertama
df['SIC_2digit'] = df['SIC'].str[:2]

# Baca file referensi SIC 2 digit
ref = pd.read_excel('2 digit.xlsx')  # File referensi

ref['SIC_2digit'] = ref['SIC_2digit'].astype(str).str.zfill(2)

# Gabungkan ke data utama berdasarkan SIC 2 digit
df = df.merge(ref, how='left', left_on='SIC_2digit', right_on='SIC_2digit')

# Tampilkan SIC 2 digit yang unik setelah merge
unique_sic_2digit = df[['SIC_2digit', 'Description']].drop_duplicates().sort_values('SIC_2digit')
print(unique_sic_2digit)

# Tampilkan hasil
print(df[['SIC', 'SIC_2digit',]])
```

```python
import pandas as pd

dataCEO = pd.read_csv('Data.csv')
dataSIC = pd.read_excel('2 digit.xlsx')

dfCEO = pd.DataFrame(dataCEO)
df2 = pd.DataFrame(dataSIC)

#Ambil 2 digit paling depan SIC nya
dfCEO['SIC'] = dfCEO['SIC'].astype(str).str.zfill(4)
dfCEO['SIC_2digit'] = dfCEO['SIC'].str[:2]

df2['SIC_2digit'] = df2['SIC_2digit'].astype(str).str.zfill(2)

gabung = pd.merge(dfCEO, df2, on='SIC_2digit', how='left')

rata_rata = gabung.groupby('SIC_1digit')[['agree', 'consc', 'extra']].mean()

print(rata_rata)
```

```python
import plotly.express as px
import matplotlib.pyplot as plt
import plotly.graph_objects as go
import pandas as pd
import os
import warnings
warnings.filterwarnings('ignore')

#Ambil Data
dataCEO = pd.read_csv('Data.csv')
dataSIC = pd.read_excel('2 digit.xlsx')

dfCEO = pd.DataFrame(dataCEO)
dfSIC = pd.DataFrame(dataSIC)

#Pastikan SICnya 4 digit dan ambil 2 digit paling depan
dfCEO['SIC'] = dfCEO['SIC'].astype(str).str.zfill(4)
dfCEO['SIC_2digit'] = dfCEO['SIC'].str[:2]

#Pastikan 2 digit dan ubah jadi string
dfSIC['SIC_2digit'] = dfSIC['SIC_2digit'].astype(str).str.zfill(2)

#Merge 2 tabel
gabung = pd.merge(dfCEO, dfSIC, on='SIC_2digit', how='left')

#Cari rata-rata tiap SIC 1 digit
rata_rata1 = gabung.groupby('SIC_1digit')[['agree', 'consc', 'extra', 'neuro', 'openn']].mean()

#Cari rata-rata tiap SIC 2 digit
rata_rata2 = gabung.groupby('SIC_2digit')[['agree', 'consc', 'extra', 'neuro', 'openn']].mean()

rata_transpose1 = rata_rata1.T
rata_transpose2 = rata_rata2.T

print(rata_transpose1)
```
