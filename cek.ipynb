```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Contoh data 60 orang × 5 aspek personality
np.random.seed(42)
data = pd.DataFrame({
    'Nama': [f'Orang_{i+1}' for i in range(60)],
    'O': np.random.rand(60),
    'C': np.random.rand(60),
    'E': np.random.rand(60),
    'A': np.random.rand(60),
    'N': np.random.rand(60)
})

# Ambil kolom aspek personality
X = data[['O','C','E','A','N']]

# Reduksi dimensi jadi 2 (PCA)
pca = PCA(n_components=2)
pca_result = pca.fit_transform(X)

# Simpan hasil
data['PC1'] = pca_result[:,0]
data['PC2'] = pca_result[:,1]

# Plot
plt.figure(figsize=(8,6))
plt.scatter(data['PC1'], data['PC2'], s=60, alpha=0.7, color='skyblue')

# Tambahkan label
for i, name in enumerate(data['Nama']):
    plt.text(data['PC1'][i]+0.01, data['PC2'][i]+0.01, name, fontsize=8)

plt.title('Visualisasi Kedekatan Personality (PCA 2D)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
import holoviews as hv
from holoviews import opts

hv.extension('bokeh')

# ==========================
# 1️⃣ Buat data acak OCEAN
# ==========================
np.random.seed(42)
names = [f'Orang_{i+1}' for i in range(10)]
aspek = ['Openness', 'Conscientiousness', 'Extraversion', 'Agreeableness', 'Neuroticism']

data = pd.DataFrame(np.random.randint(1, 8, size=(10, 5)), columns=aspek, index=names)

# ==========================
# 2️⃣ Hitung cosine similarity
# ==========================
cos_sim = cosine_similarity(data)
np.fill_diagonal(cos_sim, 0)

pairs = []
for i in range(len(names)):
    for j in range(i+1, len(names)):
        pairs.append((names[i], names[j], cos_sim[i, j]))

pairs_sorted = sorted(pairs, key=lambda x: x[2], reverse=True)[:10]
top_pairs = pd.DataFrame(pairs_sorted, columns=['Source', 'Target', 'Similarity'])

# ==========================
# 3️⃣ Buat chord diagram dengan label
# ==========================
links = top_pairs.copy()
nodes = pd.DataFrame({'Name': names})
links['Label'] = links['Similarity'].apply(lambda x: f"{x:.2f}")

# Buat objek chord
chord = hv.Chord(links).opts(
    opts.Chord(
        labels='Name',                # Label di node luar (nama orang)
        node_color='Name',            # Warna tiap node berbeda
        edge_color='Source',          # Warna garis sesuai asal node
        cmap='Category20',            # Palet warna
        edge_cmap='Category20',
        edge_alpha=0.8,
        width=800,
        height=800,
        title="Chord Diagram - Top 10 Cosine Similarity antar 10 Orang (OCEAN)",
        inspection_policy='edges',    # Hover menampilkan nilai similarity
        tools=['hover']
    )
)

# ==========================
# 4️⃣ Tambahkan hover info (keterangan)
# ==========================
# Hover akan menampilkan: Source, Target, dan nilai Similarity
hover = hv.opts.Chord(
    tools=['hover'],
    inspection_policy='edges',
    edge_hover_line_color='black',
    edge_hover_line_width=2
)

chord.opts(hover)
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
import networkx as nx
import matplotlib.pyplot as plt

# ==========================
# 1️⃣ Data acak OCEAN
# ==========================
np.random.seed(42)
names = [f'Orang_{i+1}' for i in range(10)]
aspek = ['Openness', 'Conscientiousness', 'Extraversion', 'Agreeableness', 'Neuroticism']

data = pd.DataFrame(np.random.randint(1, 8, size=(10, 5)), columns=aspek, index=names)

# ==========================
# 2️⃣ Hitung cosine similarity
# ==========================
cos_sim = cosine_similarity(data)
np.fill_diagonal(cos_sim, 0)

# Ambil pasangan dengan similarity >= 0.8
edges = []
for i in range(len(names)):
    for j in range(i+1, len(names)):
        if cos_sim[i, j] >= 0.8:
            edges.append((names[i], names[j], cos_sim[i, j]))

# ==========================
# 3️⃣ Buat graph
# ==========================
G = nx.Graph()
for name in names:
    G.add_node(name)
for (a, b, w) in edges:
    G.add_edge(a, b, weight=w)

# ==========================
# 4️⃣ Visualisasi Network
# ==========================
plt.figure(figsize=(8, 8))
pos = nx.spring_layout(G, seed=42, k=0.4)  # posisi node otomatis rapih

# Node (orang)
nx.draw_networkx_nodes(G, pos, node_size=900, node_color='skyblue', alpha=0.9)

# Edge (hubungan)
weights = [G[u][v]['weight']*5 for u,v in G.edges()]  # skala ketebalan garis
nx.draw_networkx_edges(G, pos, width=weights, alpha=0.6)

# Label nama
nx.draw_networkx_labels(G, pos, font_size=10, font_weight='bold')

# Judul
plt.title("Network Plot - Cosine Similarity antar 10 Orang (OCEAN ≥ 0.8)", fontsize=13)
plt.axis('off')
plt.show()
```

```python
# ...existing code...
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

def compute_cosine_top_pairs(X, names=None, top_n=None, threshold=None, zero_diag=True):
    """
    X: 2D array-like or DataFrame (rows = samples, cols = features)
    names: list of row names (optional). If X is DataFrame, its index is used.
    top_n: number of top pairs to return (None = all)
    threshold: min similarity to include (None = no threshold)
    zero_diag: if True set diagonal to 0 so identical rows are not treated as top pairs
    Returns: (cos_df, top_pairs_df)
    """
    if isinstance(X, pd.DataFrame):
        df = X.copy()
        vals = df.values
        if names is None:
            names = list(df.index.astype(str))
    else:
        vals = np.asarray(X)
        if names is None:
            names = [f"Item_{i}" for i in range(len(vals))]

    cos_sim = cosine_similarity(vals)
    if zero_diag:
        np.fill_diagonal(cos_sim, 0)

    cos_df = pd.DataFrame(cos_sim, index=names, columns=names)

    pairs = []
    n = len(names)
    for i in range(n):
        for j in range(i+1, n):
            s = float(cos_sim[i, j])
            if (threshold is None) or (s >= threshold):
                pairs.append((names[i], names[j], s))

    pairs_sorted = sorted(pairs, key=lambda x: x[2], reverse=True)
    if top_n is not None:
        pairs_sorted = pairs_sorted[:top_n]

    top_pairs_df = pd.DataFrame(pairs_sorted, columns=["Source", "Target", "Similarity"])
    return cos_df, top_pairs_df

# Example usage with hardcoded OCEAN data
names = ["Alice", "Bob", "Carol", "Dave", "Eve"]
data = np.array([
    [6, 5, 4, 6, 2],  # Alice
    [5, 6, 5, 5, 3],  # Bob
    [1, 2, 1, 2, 7],  # Carol
    [6, 5, 4, 6, 2],  # Dave (identical to Alice)
    [2, 3, 2, 3, 6],  # Eve (similar to Carol)
], dtype=float)

cos_matrix, top_pairs = compute_cosine_top_pairs(data, names=names, top_n=10)
print("Cosine similarity matrix:\n", cos_matrix.round(3), "\n")
print("Top pairs:\n", top_pairs)
# ...existing code...
```

```python
import pandas as pd
import numpy as np
import os
TRAIT_COLS = ['agree', 'consc', 'extra', 'neuro', 'openn']

def get_mean_sic1(dfCEO: pd.DataFrame, dfSIC: pd.DataFrame) -> pd.DataFrame:
    """
    Hitung rata-rata skor personality per SIC_1digit.
    Tambahkan Description_1 dari dfSIC jika ada.
    Selalu simpan hasil ke file Excel. Jika output_path None, gunakan 'mean_sic1.xlsx' di CWD.

    Parameters:
    - dfCEO, dfSIC: input dataframes
    - output_path: optional path untuk menyimpan excel (str). Jika None -> 'mean_sic1.xlsx'.
    Returns:
    - rata1: DataFrame hasil rata-rata (juga disimpan ke Excel)
    """
    from pathlib import Path

    dfCEO = dfCEO.copy()
    dfSIC = dfSIC.copy()

    print(dfCEO)

    # Pastikan format kolom kunci seragam
    dfCEO['SIC_1digit'] = dfCEO['SIC_1digit'].astype(str).str.strip()
    dfSIC['SIC_1digit'] = dfSIC['SIC_1digit'].astype(str).str.strip()

    # Hitung rata-rata berdasarkan SIC_1digit
    rata1 = dfCEO.groupby('SIC_1digit')[TRAIT_COLS].mean(numeric_only=True).reset_index()

    # Gabungkan dengan deskripsi (jika ada di dfSIC)
    if 'Description_1' in dfSIC.columns:
        desc = dfSIC[['SIC_1digit', 'Description_1']].drop_duplicates()
        rata1 = rata1.merge(desc, on='SIC_1digit', how='left')
    else:
        # fallback kalau description belum ada
        rata1['Description_1'] = rata1['SIC_1digit']

    # Tentukan path output dan simpan ke Excel (menggunakan openpyxl)
    output_path = Path.cwd() / 'mean_sic1.xlsx'

    try:
        rata1.to_excel(output_path, index=False, engine='openpyxl')
    except Exception as e:
        # Jika gagal menulis, raise supaya pemanggil tahu ada masalah I/O
        raise IOError(f"Gagal menyimpan hasil ke Excel di '{output_path}': {e}")

    return rata1

dfCEO = pd.read_csv('data/Data.csv')
dfSIC = pd.read_excel('data/2 digit.xlsx')
get_mean_sic1(dfCEO, dfSIC)
```
